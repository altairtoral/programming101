```python
#Calculate Simple interest
def simple_interest(p, r, t, show=False):
    '''p: float
           Ammount in €#
        r: float
            Interest rate in %
        t: int
            Time in years
    '''
    si=(p*r*t)/100
    a= p+si
    if show:
        print('SIMPLE INTEREST')
        print('Amount: ', p,'€') #in Euros
        print("Rate of Interest (Per Annum)", r,'%') #In %
        print("Time (In Years): ",t,'years')
        print("Final Amount: ", a,'€')
        print("Simple Interest: ", si,'€')
    return si, a
```

```python
#compound interest
def compound_interest(p,r,t, show=False):
    a= p*((1+r/100)**t)
    ci= a-p
    if show:
        print ('COMPOUND INTEREST')
        print('Amount: ', p,'€')
        print("Rate of Interest (Per Annum)", r, '%')
        print("Time (In Years): ",t,'years')
        print("Final Amount: ", a,'€')
        print("Compound Interest: ", ci,'€')
    return ci,a
```

```python
# Purchasing power
def pur_power(p,i,t,show=False):
    '''This function calculates the change in purchase power due to inflation.
        You will get the final value "a" of the initial capital "p" after "t" years of devaluation due to inflation "i"'''
    a= p* ((100/(100+i))**t) # Calculate final amount
    if show:
        print ('PURCHASING POWER')
        print("Initial Amount: ",p)
        print("Annual Inflation Rate: ",i) #Also discount rate
        print("Time in years:", t)
        print("Equivalent purchasing value after",t,"years of inflation: ", a)
    return a # Return final value 
```

```python
#compound annual growth rate
def CAGR(p,a,t,show=False):
    t_inv=1/t
    cagr_rate= (((a/p)**t_inv)  -1)*100 #calculate compund annual growth rate
    if show:
        print ('COMPOUND ANUAL GROWTH RATE')
        print("Initial value of money: ",p)
        print('Final value of money: ',a)
        print("Time in years: ", t)
        print("Compound Annual Growth Rate (CAGR): ", cagr_rate)
    return cagr_rate
```

```python
#monthly EMI calculation
def EMI(p,r,n,show=False):
    r_mon=  r/(12*100) #Monthly rate
    emi = p * r_mon * ((1+r_mon)**n)/((1+r_mon)**n - 1)
    if show:
        print("Total Loan Amount: ",p)
        print("Rate of Interest: ",r)
        print("Number of installments: ",n)
        print("EMI for",n,"months: ",emi )
    return emi
```
