```python
def lineq(x,slope,intercept):
    return slope*x+intercept
```

```python
class Person:
    def __init__(self, name, birthday, gender, height, weight, hairColor):
        self.name=name
        self.birthday=birthday
        self.gender= gender
        self.height = height
        self.weight= weight
        self.hairColor=hairColor
                
    def getBirthdayDate(self):
        birthDate=datetime.strptime(self.birthday,"%d.%m.%Y")
        return birthDate
        #return self.birthday.strftime('%d.%m.%Y')
    
    def getRetirementDate(self, retirementAge=67):
        #if retirementAge==Null:
        #    retirementAge=67
        d=datetime.today()
        thisYear=d.year
        birthDate=self.getBirthdayDate()
        age=thisYear-birthDate.year
        yearstoWork=retirementAge-age
        retirementYear=thisYear+yearstoWork
        print (self.name, "you have to work", yearstoWork, "years before retirement")
        print ("you are going to retire in", retirementYear)
        return retirementYear, yearstoWork
```

```python
tona=Person ("Tona", "07.08.1982","M", 1.72,75,"black")
print (tona.name)
print (tona.birthday)
print (tona.gender)
(RDate,ToWork)=tona.getRetirementDate()
#a=tona.getBirthdayDate()
#retirementAge=67
#Age=thisYear-a.year
#print (tona.name, "you are",Age, "years old")
#yearstoWork=retirementAge-Age
#print (tona.name,"you have to work",yearstoWork,"years")
#print (tona.birthday())
```

```python
#inflation=pd.read_csv('InflationData/InflationData.csv')
inflationde=pd.read_csv('InflationData/adaptedInflationDatade.csv')
#inflation
```

```python
inflationde.shape
#inflation.describe()
#inflationde
inflationde.columns
xtotal= np.array(inflationde['Year'])
ytotal= np.array(inflationde['Inflation'])
x= xtotal[40:] #only data from year 2000
y= ytotal[40:]
```

```python
#inflation.plot("Germany")
#inflation["Country Name"]
#DE=inflation.loc[55]
#print (inflation[55:56])
#print (inflation.iloc[55])
#print (inflation.loc[55])
#DE.plot(4:)
```

```python
#plt.plot (inflation)
#DE=DE.iloc[4:]
DE=inflationde
plt.plot (x,y, 'r-', label="Germany")
plt.title ("Inflation in Germany from 1960 to 2020")
plt.ylabel('%')
#plt.xticks
plt.show()
```

```python
#Applying a linear fit
fit=np.polyfit(x,y,1)
slope=fit[0]
intercept=fit[1]
bestfit=slope*x+intercept # y=m*x+b
```

```python
print (fit)
```

```python
#Inflation by retirement

#prepare data
#y2=np.array[lineq(i,slope,intercept) for i in range(ToWork)]

sizeof2ndarray= RDate-x[x.size-1]
x2=np.linspace(x[x.size-1]+1,RDate,sizeof2ndarray)
y2=np.zeros(sizeof2ndarray)
x=np.concatenate([x,x2]) #extending x axis to date of retirement
y=np.concatenate([y,y2]) #filling y with zeros


#Interpolate the best fit to date of retirement

interpolation=np.array([lineq(i,slope,intercept) for i in x2])
bestfit=np.concatenate([bestfit,interpolation])
#for i in x2
#ax.plot(x,y2,color='g',label='in the future')
```

```python

#Plotting
fig= plt.figure()
ax=fig.subplots()
ax.plot(x,bestfit, color='r', alpha=0.5, label='Best Fit')
ax.plot(x,y,color='b', label='Inflation')
ax.set_title('Linear regration of Inflation')
plt.ylabel('%')
ax.legend()
plt.show()
```
