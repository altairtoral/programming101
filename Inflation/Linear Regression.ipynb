```python
import numpy as np
from numpy import random
import matplotlib.pyplot as plt
```

```python
#Creating random data
x= np.linspace(0,60,60)
y= np.array([random.randint(i-2, i+2) for i in x])
```

```python
#Applying a linear fit
fit=np.polyfit(x,y,1)
slope=fit[0]
intercept=fit[1]
bestfit=slope*x+intercept # y=m*x+b
```

```python
#Plotting
fig= plt.figure()
ax=fig.subplots()
ax.plot(x,bestfit, color='r', alpha=0.5, label='Best Fit')
ax.scatter(x,y,s=5,color='b', label='Data points')
ax.set_title('Linear Fit Example')
ax.legend()
plt.show()
```
